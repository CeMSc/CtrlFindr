# CtrlFindr

Run a CtrlFindr project from a notebook. The same steps are available on the command line:

```
python -m ctrlfindr run --config path/to/ctrlfindr.toml
```

For long runs on a JupyterLab server, use the command line in a JupyterLab terminal: it keeps running if the browser tab is closed.

Install once, from the repository folder: `pip install -r requirements.lock`, then `pip install -e . --no-deps`.

````python
from pathlib import Path

from ctrlfindr import analyze, ingest, load_config
from ctrlfindr.common import setup_logging

# Path to the project's settings file (edit this).
CONFIG = Path("../examples/example_project/ctrlfindr.toml")

cfg = load_config(CONFIG)
setup_logging(log_file=cfg.paths.output / "ctrlfindr.log")
````

## 1. Ingest

Reads the text files into the database. Only new or changed files are processed, so re-running is cheap.

````python
ingest.run(cfg)
````

## 2. Analyze

Validates the assessment framework, codes every sentence and writes the five TSV files plus `run_metadata.json` to the output folder. Re-run this cell after editing the framework; there is no need to ingest again.

````python
outputs = analyze.run(cfg)
outputs["results_document"]
````
